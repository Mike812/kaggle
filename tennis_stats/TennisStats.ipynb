### Read tennis datasets from kaggle as dataframes

```python
data_path = "data/"
players_and_years = pd.read_csv(data_path+"players(man).csv")
players_tournament_results = pd.read_csv(data_path+"players_tournament(man).csv")
players_tournament_detailed = pd.read_csv(data_path+"raw_kaggle.csv")
players_tournament_serve_data = pd.read_csv(data_path+"serve_kaggle.csv")
players_tournament_return_data = pd.read_csv(data_path+"return_kaggle.csv")
```

```python
def query_number_of_matches(more_than_no_of_matches):
    result = players_and_years[players_and_years.number_of_matches > more_than_no_of_matches].sort_values(by="number_of_matches", ascending=False).reset_index(drop=True)
    return result

def query_number_of_years(more_than_no_of_years):
    result = players_and_years[players_and_years.number_of_years > more_than_no_of_years].sort_values(by="number_of_matches", ascending=False).reset_index(drop=True)
    return result

def query_atp_finalists():
    return players_tournament_results[players_tournament_results["number_of_rounds"] > 5].sort_values(by="year", ascending=False).reset_index(drop=True)
    

def query_stats_file(stat_file, tennis_player, tournament, year, agg_function=None):
    result = stat_file[stat_file.Name.str.contains(tennis_player) & stat_file.Tournament.str.contains(tournament) 
    & stat_file.Date.str.contains(year)]

    if agg_function:
        result = agg_function(result)
    
    return result

def agg_players_tournament_detailed(df):
    result = df[["Name", "TP", "Aces", "DFs", "SP", "1SP", "2SP", "vA"]].groupby("Name").mean()
    return result
```
