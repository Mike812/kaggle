```python
import pandas as pd
```

### Read tennis datasets from kaggle as dataframes

```python
data_path = "data/"
players_and_years = pd.read_csv(data_path+"players(man).csv")
players_tournament_results = pd.read_csv(data_path+"players_tournament(man).csv")
players_tournament_detailed = pd.read_csv(data_path+"raw_kaggle.csv")
players_tournament_serve_data = pd.read_csv(data_path+"serve_kaggle.csv")
players_tournament_return_data = pd.read_csv(data_path+"return_kaggle.csv")
```

```python
players_and_years.info()
```

```python
players_and_years.head()
```

```python
players_and_years["number_of_years"] = players_and_years["playing_years"].str.count(',') + 1
```

```python
players_and_years.sort_values(by="number_of_matches", ascending=False).reset_index(drop=True)
```

```python
players_tournament_results["number_of_rounds"] = players_tournament_results["rounds"].str.count(",") + 1
```

```python
players_tournament_results.sort_values(by="number_of_rounds", ascending=False).reset_index(drop=True)
```
